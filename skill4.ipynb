# Step 11: Check CUDA and GPU Availability

CUDA is a parallel computing platform developed by NVIDIA.

PyTorch uses CUDA to execute deep learning operations on NVIDIA GPUs.

In this step, we check whether CUDA is available in our Google Colab runtime.

```python
import torch

# Check CUDA availability
cuda_available = torch.cuda.is_available()

print("CUDA Available:", cuda_available)

# Check CUDA version
print("CUDA Version:", torch.version.cuda)

# Check GPU device
if cuda_available:
    print("GPU Device:", torch.cuda.get_device_name(0))
else:
    print("GPU is not available. PyTorch will use CPU.")
```

# Step 12: PyTorch Tensor Creation and Device Execution

Tensors are the basic data structures in PyTorch.

They can store numerical values and perform mathematical operations.

In this step:
1. Create a tensor.
2. Perform a mathematical operation on CPU.
3. Check whether GPU execution is available.
4. Perform the same operation on GPU if CUDA is available.

```python
import torch

# Create a tensor
x = torch.tensor([10, 20, 30, 40])

# Perform computation on CPU
cpu_result = x * 2

print("Original Tensor:", x)
print("CPU Result:", cpu_result)
print("CPU Device:", x.device)

# Perform computation on GPU if available
if torch.cuda.is_available():
    gpu_tensor = x.to("cuda")
    gpu_result = gpu_tensor * 2

    print("\nGPU Result:", gpu_result)
    print("GPU Device:", gpu_tensor.device)
else:
    print("\nCUDA is not available.")
    print("GPU computation skipped.")
```

# Step 13: Automatic Differentiation and Gradients

Automatic differentiation is a feature in PyTorch that calculates gradients automatically.

Gradients help neural networks update their parameters during training.

In this experiment:
1. Create a tensor with requires_grad=True.
2. Define a mathematical function.
3. Calculate the output.
4. Use backward() to calculate the gradient.

```python
import torch

# Create a tensor that tracks gradients
x = torch.tensor(2.0, requires_grad=True)

# Define a mathematical function
y = x**2 + 3*x + 1

# Display the output
print("Input x:", x.item())
print("Output y:", y.item())

# Calculate gradient
y.backward()

# Display gradient
print("Gradient of y with respect to x:", x.grad.item())
```

# Step 14: Generate PyTorch and CUDA Environment Report

An environment report provides information about the development environment.

It includes:
1. Python version
2. PyTorch version
3. CUDA availability
4. CUDA version
5. GPU device information
6. CPU or GPU execution status

This report helps verify the development environment.

```python
import sys
import torch
import platform

# Collect environment information
environment_report = {
    "Python Version": sys.version,
    "Operating System": platform.platform(),
    "PyTorch Version": torch.__version__,
    "CUDA Available": torch.cuda.is_available(),
    "CUDA Version": torch.version.cuda,
    "GPU Device": (
        torch.cuda.get_device_name(0)
        if torch.cuda.is_available()
        else "Not Available"
    ),
    "Execution Device": (
        "GPU" if torch.cuda.is_available() else "CPU"
    )
}

# Display the environment report
print("PYTORCH AND CUDA ENVIRONMENT REPORT")
print("-" * 45)

for key, value in environment_report.items():
    print(f"{key}: {value}")
```

# Step 15: Save Environment Report as JSON

JSON is a file format used to store structured information.

In this step:
1. Convert the environment report into JSON format.
2. Save it as environment_report.json.
3. Read the saved file.
4. Display its contents to verify successful saving.

```python
import json
import os

# Define the report file path
report_path = os.path.join(project_path, "reports", "environment_report.json")

# Save the environment report
with open(report_path, "w") as file:
    json.dump(environment_report, file, indent=4)

print("Environment report saved successfully!")
print("File Location:", report_path)

# Read and display the saved report
with open(report_path, "r") as file:
    saved_report = json.load(file)

print("\nSaved JSON Report:")
print(json.dumps(saved_report, indent=4))
```
